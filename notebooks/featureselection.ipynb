```python
"""
════════════════════════════════════════════════════════════════════════════════
ANALISIS FITUR UNTUK PREDIKSI SOLAR IRRADIANCE
════════════════════════════════════════════════════════════════════════════════
"""

import pandas as pd
import numpy as np

# Load data
data_csv = """date,station_id,station_name,region_name,province_id,province_name,latitude,longitude,Tavg,Tx,Tn,ss,RH_avg,RR,solar_radiation_MJ,cloud_cover_pct,sunshine_hours
2010-01-01,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,27.1,30.2,21.4,0.5,82.0,9.0,19.72,94.0,10.827569444444444
2010-01-02,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,25.7,29.6,21.0,0.2,95.0,24.0,13.35,100.0,6.938352777777777
2010-01-03,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,24.5,26.8,20.2,0.0,98.0,63.0,10.85,100.0,4.601330555555556
2010-01-04,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,25.8,29.2,21.0,0.1,90.0,0.0,9.43,100.0,5.436005555555555
2010-01-05,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.7,30.0,21.2,0.4,90.0,2.0,18.27,90.0,9.877661111111111
2010-01-06,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.1,30.0,21.2,0.3,93.0,11.0,15.75,99.0,8.097855555555554
2010-01-07,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,25.4,29.0,21.4,0.1,96.0,2.0,13.71,99.0,8.240725
2010-01-08,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.8,29.8,21.8,0.6,91.0,3.0,16.05,95.0,7.7074083333333325
2010-01-09,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.7,30.2,21.8,0.7,92.0,3.0,19.09,93.0,10.778705555555554
2010-01-10,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,27.1,30.2,20.0,0.6,88.0,28.0,20.43,58.0,10.830494444444444
2010-01-11,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,25.7,30.2,20.0,0.6,92.0,4.0,19.56,49.0,10.766919444444445
2010-01-12,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,27.3,30.2,23.4,0.6,85.0,0.0,20.07,35.0,10.868036111111111
2010-01-13,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.9,30.2,21.8,0.2,82.0,0.0,20.47,80.0,10.764802777777778
2010-01-14,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.3,30.0,22.0,0.6,87.0,3.0,18.47,86.0,9.60868611111111
2010-01-15,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.4,29.8,22.4,0.4,88.0,1.0,21.03,56.0,10.8215
2010-01-16,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,27.4,29.8,22.8,0.6,82.0,0.0,20.76,31.0,10.979022222222223
2010-01-17,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,27.5,30.8,22.2,0.6,80.0,0.0,20.33,37.0,10.898402777777777
2010-01-18,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,26.7,30.4,21.8,0.6,89.0,0.0,18.69,69.0,10.763175
2010-01-19,96001,Stasiun Meteorologi Maimun Saleh,Kota Sabang,1,Nanggroe Aceh Darussalam,5.87655,95.33785,25.9,29.8,20.4,0.2,95.0,22.0,19.14,76.0,10.8386"""

from io import StringIO
df = pd.read_csv(StringIO(data_csv))
df['date'] = pd.to_datetime(df['date'])

# Calculate correlations dengan target (solar_radiation_MJ)
target = 'solar_radiation_MJ'
numeric_cols = ['Tavg', 'Tx', 'Tn', 'ss', 'RH_avg', 'RR', 'cloud_cover_pct', 'sunshine_hours']
correlations = df[numeric_cols + [target]].corr()[target].drop(target)

# ════════════════════════════════════════════════════════════════════════════
# CREATE FEATURE IMPORTANCE TABLE
# ════════════════════════════════════════════════════════════════════════════

feature_analysis = []

# TARGET VARIABLE
feature_analysis.append({
    'No': '🎯',
    'Feature': 'solar_radiation_MJ',
    'Type': 'TARGET',
    'Importance': '⭐⭐⭐⭐⭐',
    'Correlation': 'N/A',
    'Kategori': '🎯 TARGET',
    'Pakai?': '✅ YES',
    'Penjelasan': 'INI YANG MAU DIPREDIKSI! Output model.'
})

# IDENTIFIERS (tidak untuk training)
identifiers = [
    ('date', 'Identifier', 'Tanggal record. Bisa extract jadi temporal features (month, day_of_year, season)'),
    ('station_id', 'Identifier', 'ID stasiun. Untuk grouping, bukan fitur langsung.'),
    ('station_name', 'Identifier', 'Nama stasiun. Informasi saja, bukan fitur numerik.'),
    ('region_name', 'Identifier', 'Nama wilayah. Bisa di-encode jadi categorical tapi biasanya gak terlalu berguna.'),
    ('province_id', 'Identifier', 'ID provinsi. Bisa jadi categorical feature kalau mau model per-region.'),
    ('province_name', 'Identifier', 'Nama provinsi. Sama seperti province_id.'),
    ('latitude', 'Geographic', 'Latitude stasiun. Bisa berguna untuk spatial features tapi korelasi lemah.'),
    ('longitude', 'Geographic', 'Longitude stasiun. Sama dengan latitude.'),
]

for feat, typ, explanation in identifiers:
    feature_analysis.append({
        'No': '📋',
        'Feature': feat,
        'Type': typ,
        'Importance': '⭐',
        'Correlation': 'N/A',
        'Kategori': '❌ SKIP',
        'Pakai?': '❌ NO',
        'Penjelasan': explanation
    })

# PREDICTORS - sorted by correlation
predictor_info = {
    'cloud_cover_pct': {
        'desc': 'Cloud cover %',
        'importance': '⭐⭐⭐⭐⭐',
        'kategori': '🔥 CRITICAL',
        'use': '✅ YES',
        'explanation': 'PALING PENTING! Cloud cover menutupi matahari = solar radiation turun drastis. Korelasi negatif sangat kuat.'
    },
    'ss': {
        'desc': 'Sunshine duration (jam)',
        'importance': '⭐⭐⭐⭐⭐',
        'kategori': '🔥 CRITICAL',
        'use': '✅ YES',
        'explanation': 'SANGAT PENTING! Durasi sinar matahari berbanding lurus dengan solar radiation. Fitur terkuat kedua.'
    },
    'sunshine_hours': {
        'desc': 'Sunshine dari Open-Meteo',
        'importance': '⭐⭐⭐⭐',
        'kategori': '✅ IMPORTANT',
        'use': '⚠️ OPTIONAL',
        'explanation': 'Sama dengan ss tapi dari source berbeda. Bisa di-drop kalau redundant dengan ss atau di-average.'
    },
    'Tavg': {
        'desc': 'Temperature rata-rata (°C)',
        'importance': '⭐⭐⭐⭐',
        'kategori': '✅ IMPORTANT',
        'use': '✅ YES',
        'explanation': 'Suhu tinggi biasanya = cuaca cerah = solar radiation tinggi. Korelasi moderate-strong.'
    },
    'Tx': {
        'desc': 'Temperature maksimum (°C)',
        'importance': '⭐⭐⭐',
        'kategori': '✅ IMPORTANT',
        'use': '✅ YES',
        'explanation': 'Temp max indikator puncak radiasi. Korelasi moderate. Bisa combine dengan Tavg jadi feature baru (temp_range = Tx - Tn).'
    },
    'Tn': {
        'desc': 'Temperature minimum (°C)',
        'importance': '⭐⭐',
        'kategori': '⚠️ MODERATE',
        'use': '⚠️ OPTIONAL',
        'explanation': 'Temp min korelasi lemah dengan solar. Bisa dipakai tapi gak terlalu pengaruh. Lebih berguna untuk feature engineering (temp_range).'
    },
    'RH_avg': {
        'desc': 'Relative Humidity (%)',
        'importance': '⭐⭐⭐',
        'kategori': '✅ IMPORTANT',
        'use': '✅ YES',
        'explanation': 'Humidity tinggi = banyak water vapor = mengurangi transmisi radiasi. Korelasi negatif moderate.'
    },
    'RR': {
        'desc': 'Rainfall (mm)',
        'importance': '⭐⭐',
        'kategori': '⚠️ MODERATE',
        'use': '⚠️ OPTIONAL',
        'explanation': 'Hujan = clouds = radiasi turun. Tapi korelasi lemah karena hujan gak selalu sepanjang hari. Bisa dibuat binary feature (is_rainy).'
    },
}

# Add predictors sorted by absolute correlation
for feat in correlations.abs().sort_values(ascending=False).index:
    if feat in predictor_info:
        info = predictor_info[feat]
        corr_val = correlations[feat]
        
        feature_analysis.append({
            'No': len([x for x in feature_analysis if x['Type'] == 'Predictor']) + 1,
            'Feature': feat,
            'Type': 'Predictor',
            'Importance': info['importance'],
            'Correlation': f"{corr_val:+.3f}",
            'Kategori': info['kategori'],
            'Pakai?': info['use'],
            'Penjelasan': info['explanation']
        })

# Convert to DataFrame
analysis_df = pd.DataFrame(feature_analysis)

# ════════════════════════════════════════════════════════════════════════════
# PRINT RESULTS
# ════════════════════════════════════════════════════════════════════════════

print("\n" + "="*120)
print(" " * 40 + "ANALISIS FITUR UNTUK PREDIKSI SOLAR IRRADIANCE")
print("="*120)

print("\n📊 TABEL LENGKAP SEMUA FITUR:")
print("-"*120)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
pd.set_option('display.max_colwidth', 50)
print(analysis_df.to_string(index=False))

# ════════════════════════════════════════════════════════════════════════════
# SUMMARY
# ════════════════════════════════════════════════════════════════════════════

print("\n" + "="*120)
print("📌 SUMMARY REKOMENDASI")
print("="*120)

print("\n🔥 FITUR WAJIB PAKAI (CRITICAL):")
critical = analysis_df[analysis_df['Kategori'] == '🔥 CRITICAL']
for _, row in critical.iterrows():
    print(f"   ✅ {row['Feature']:20s} → {row['Penjelasan'][:90]}")

print("\n✅ FITUR PENTING (IMPORTANT):")
important = analysis_df[analysis_df['Kategori'] == '✅ IMPORTANT']
for _, row in important.iterrows():
    print(f"   ✅ {row['Feature']:20s} → {row['Penjelasan'][:90]}")

print("\n⚠️  FITUR OPTIONAL (MODERATE):")
moderate = analysis_df[analysis_df['Kategori'] == '⚠️ MODERATE']
for _, row in moderate.iterrows():
    print(f"   ⚠️  {row['Feature']:20s} → {row['Penjelasan'][:90]}")

print("\n❌ SKIP (Identifiers - Jangan masuk model sebagai fitur langsung):")
skip = analysis_df[analysis_df['Kategori'] == '❌ SKIP']
for _, row in skip.iterrows():
    print(f"   ❌ {row['Feature']:20s} → {row['Penjelasan'][:90]}")

# ════════════════════════════════════════════════════════════════════════════
# CORRELATION ANALYSIS
# ════════════════════════════════════════════════════════════════════════════

print("\n" + "="*120)
print("📈 KORELASI DENGAN SOLAR_RADIATION_MJ (sorted by strength)")
print("="*120)

corr_sorted = correlations.abs().sort_values(ascending=False)
for feat, corr_abs in corr_sorted.items():
    corr_actual = correlations[feat]
    
    # Strength indicator
    if corr_abs > 0.7:
        strength = "🔥 SANGAT KUAT"
        emoji = "🔥"
    elif corr_abs > 0.5:
        strength = "✅ KUAT"
        emoji = "✅"
    elif corr_abs > 0.3:
        strength = "⚠️  MODERATE"
        emoji = "⚠️"
    else:
        strength = "❌ LEMAH"
        emoji = "❌"
    
    # Direction
    direction = "📈 POSITIF" if corr_actual > 0 else "📉 NEGATIF"
    
    print(f"   {emoji} {feat:20s}: {corr_actual:+.3f}  ({strength}, {direction})")

# ════════════════════════════════════════════════════════════════════════════
# FINAL RECOMMENDATION
# ════════════════════════════════════════════════════════════════════════════

print("\n" + "="*120)
print("🎯 REKOMENDASI FINAL UNTUK MODEL")
print("="*120)

print("""
FITUR INPUT (X) YANG DIPAKAI:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

1. 🔥 cloud_cover_pct       → Fitur TERKUAT! Harus ada!
2. 🔥 ss                     → Sunshine duration, sangat penting!
3. ✅ Tavg                   → Temperature rata-rata
4. ✅ Tx                     → Temperature maksimum
5. ✅ RH_avg                 → Relative humidity

OPTIONAL (boleh dipakai, boleh di-skip):
6. ⚠️  sunshine_hours        → Redundant dengan ss, pilih salah satu
7. ⚠️  Tn                    → Bisa combine dengan Tx jadi temp_range = Tx - Tn
8. ⚠️  RR                    → Bisa dibuat binary: is_rainy = RR > 1

TARGET (Y):
🎯 solar_radiation_MJ        → Ini yang mau diprediksi!

JANGAN DIPAKAI SEBAGAI FITUR LANGSUNG:
❌ date, station_id, station_name, region_name, province_id, province_name, latitude, longitude
   (Ini identifier/metadata, bukan fitur numerik yang relevan)

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

FEATURE ENGINEERING TAMBAHAN:
✅ month, season              → Extract dari date
✅ is_monsoon                 → Boolean (Oct-Mar di Indonesia)
✅ temp_range = Tx - Tn       → Range suhu harian
✅ is_rainy = RR > 1          → Binary rainfall indicator
✅ lag features               → solar_radiation_lag_1, cloud_cover_lag_1
✅ rolling mean               → solar_radiation_rolling_mean_7

MODEL RECOMMENDATIONS:
✅ CNN-LSTM (sequence 7-14 days) → Buat time series
✅ Transformer                   → Attention mechanism
✅ XGBoost/RandomForest          → Baseline comparison

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
""")

print("\n" + "="*120)
print("✅ ANALISIS SELESAI!")
print("="*120)
```
